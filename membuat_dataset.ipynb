```python
# Import Library
import pandas as pd
import numpy as np
```

```python
# Mengatur random seed agar hasil dapat direplikasi
np.random.seed(42)

n_samples = 200

nilai_tugas = np.random.randint(50, 100, n_samples)
nilai_uts = np.random.randint(45, 100, n_samples)
nilai_uas = np.random.randint(40, 100, n_samples)
kehadiran = np.random.randint(60, 100, n_samples)
jumlah_absen = np.random.randint(0, 12, n_samples)
```

```python
# Menampilakan nilai random seed data tiap kategori
print("Nilai Tugas:", nilai_tugas)
print("Nilai UTS:", nilai_uts)
print("Nilai UAS:", nilai_uas)
print("Kehadiran:", kehadiran)
print("Jumlah Absen:", jumlah_absen)
```

```python
# Aturan logis pembuatan label target: Lulus (1) atau Tidak Lulus (0)
# Nilai akhir rata-rata terbobot
nilai_akhir = (nilai_tugas * 0.2) + (nilai_uts * 0.3) + (nilai_uas * 0.5)

status_lulus = []
for i in range(n_samples):
    if nilai_akhir[i] >= 65 and kehadiran[i] >= 75 and jumlah_absen[i] <= 5:
        status_lulus.append("Lulus")
    else:
        status_lulus.append("Tidak Lulus")

df = pd.DataFrame({
    'Siswa_ID': [f'SISWA_{i+1:03d}' for i in range(n_samples)],
    'Nilai_Tugas': nilai_tugas,
    'Nilai_UTS': nilai_uts,
    'Nilai_UAS': nilai_uas,
    'Persentase_Kehadiran': kehadiran,
    'Jumlah_Absen': jumlah_absen,
    'Status_Lulus': status_lulus
})

# Menampilkan 5 data pertama
print("--- 5 Data Pertama Dataset ---")
display(df.head())
```

```python
# Cell 3: Pemeriksaan dan Informasi Dataset
print("=== INFORMASI DATASET ===")
df.info()

print("\n=== PEMERIKSAAN MISSING VALUES ===")
print(df.isnull().sum())

print("\n=== DISTRIBUSI TARGET (STATUS LULUS) ===")
print(df['Status_Lulus'].value_counts())

print("\n=== STATISTIK DESKRIPTIF FITUR ===")
display(df.describe())
```

```python
# Simpan ke Excel
df.to_excel('dataset_kelulusan.xlsx', index=False)
print("Dataset berhasil dibuat: dataset_kelulusan.xlsx")
```
